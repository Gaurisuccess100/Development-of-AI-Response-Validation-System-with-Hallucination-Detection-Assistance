Write Python code to count the number of vowels in a string.

```python
text = "Python Programming"
vowels = "aeiou"

count = 0

for char in text.lower():
    if char in vowels:
        count += 1

print("Number of vowels:", count)
```

Create a Pandas DataFrame containing student names and marks.

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C"],
    "Marks": [80, 75, 90]
}

df = pd.DataFrame(data)

print(df)
```

Find the average marks and maximum marks using Pandas.

```python
print("Average:", df["Marks"].mean())
print("Maximum:", df["Marks"].max())
```

Create a DataFrame with missing values and count the missing values in each column.

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C"],
    "Marks": [80, None, 90]
}

df = pd.DataFrame(data)

print(df.isnull().sum())
```

Select students whose marks are greater than 80.

```python
result = df[df["Marks"] > 80]

print(result)
```

Calculate the average salary for each department using Pandas.

```python
data = {
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 60000, 45000]
}

df = pd.DataFrame(data)

print(df.groupby("Department")["Salary"].mean())
```

Create a simple line graph using Matplotlib.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.plot(x, y)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales Trend")
plt.show()
```

Create a bar chart using Seaborn.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Subject": ["Python", "SQL", "ML"],
    "Marks": [85, 90, 80]
}

df = pd.DataFrame(data)

sns.barplot(x="Subject", y="Marks", data=df)
plt.show()
```

Sort a DataFrame according to marks in descending order.

```python
data = {
    "Name": ["A", "B", "C"],
    "Marks": [70, 90, 80]
}

df = pd.DataFrame(data)

print(df.sort_values("Marks", ascending=False))
```

Calculate the correlation between two numerical columns using Pandas.

```python
data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 60, 80, 100]
}

df = pd.DataFrame(data)

print(df["Hours"].corr(df["Marks"]))
```

Train a simple linear regression model using Scikit-learn.

```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4]]
y = [10, 20, 30, 40]

model = LinearRegression()
model.fit(X, y)

print(model.predict([[5]]))
```

Create a Pandas DataFrame, calculate average marks, find students who scored above 80, and display the result.

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [75, 85, 90, 70]
}

df = pd.DataFrame(data)

print("Average:", df["Marks"].mean())

print("\nStudents above 80:")
print(df[df["Marks"] > 80])
```
